```python
# Data manipulation
import pandas as pd
import numpy as np

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns
```

### Dataset Loading

The dataset is loaded from a CSV file containing short text statements labeled with
mental health categories. No preprocessing is applied at this stage to preserve the
original data distribution.


```python
# Load the dataset
df = pd.read_csv("../data/raw/sentiment_mental_health.csv")
```

### Dataset Structure Inspection

- Column names and data types

```python
# Display first few rows
df.head()
```

### Missing Value Analysis

Missing values can negatively impact text vectorization and model training.
This step verifies whether data cleaning or imputation is required.


```python
# Check for missing values
df.isnull().sum()
```

### Class Distribution Analysis

Understanding label distribution is critical for:
- Selecting appropriate evaluation metrics
- Interpreting model performance
- Identifying potential class imbalance issues


```python
# Count samples per class
df["status"].value_counts()
```

```python
# Visualize class distribution
plt.figure(figsize=(12, 4))
sns.countplot(x="status", data=df)
plt.title("Distribution of Mental Health Labels")
plt.xlabel("Label")
plt.ylabel("Number of Samples")
plt.show()
```

### Text Length Analysis

Analyzing text length provides insight into whether:
- Traditional NLP techniques are appropriate
- Long-sequence models are necessary

Short text lengths typically favor vector-based models such as TF-IDF.

### Summary of Step 1 Findings

- The dataset consists of short, informal text statements labeled by mental health category
- No missing values were detected
- Class imbalance is present across labels
- Text length is relatively short, supporting the use of TF-IDF-based models

These findings motivate the preprocessing and modeling choices in the next step.
